```python
import sys
import subprocess
import importlib

def ensure_package(pkg_name):
    try:
        importlib.import_module(pkg_name)
    except ImportError:
        subprocess.check_call([sys.executable, "-m", "pip", "install", pkg_name])

# Ensure required packages are available
ensure_package("nbformat")
ensure_package("nbconvert")

import nbformat
from nbconvert import HTMLExporter
from IPython.display import FileLink, display
import os

nb_path = "Example2.ipynb"   # change if your notebook has a different name
out_path = os.path.splitext(nb_path)[0] + ".html"

# Read notebook and convert to HTML
nb = nbformat.read(nb_path, as_version=4)
html_exporter = HTMLExporter()
body, resources = html_exporter.from_notebook_node(nb)

# Write HTML output
with open(out_path, "w", encoding="utf-8") as f:
    f.write(body)

print(f"Wrote {out_path}")
display(FileLink(out_path))
```

```python
import sys
import os
import importlib
import zipfile

# Ensure packages are available
def ensure_pkg(name):
    try:
        importlib.import_module(name)
    except ImportError:
        print(f"Installing {name}...")
        import subprocess
        subprocess.check_call([sys.executable, "-m", "pip", "install", name])

ensure_pkg("nbformat")
ensure_pkg("nbconvert")

import nbformat
from nbconvert import HTMLExporter
from IPython.display import FileLink, display

# Paths / names
nb_path = "Example3.ipynb"
html_path = os.path.splitext(nb_path)[0] + ".html"
zip_path = "notebook_outputs.zip"
files_to_include = [
    nb_path,
    html_path,
    "students.csv",
    "students.xlsx",
    "students.pkl",
    "scores_chart.png",
    "plotly_fig.html"
]

# Convert notebook to HTML (only if notebook exists)
if not os.path.exists(nb_path):
    print(f"Notebook not found: {nb_path}. Aborting conversion.")
else:
    try:
        nb = nbformat.read(nb_path, as_version=4)
        html_exporter = HTMLExporter()
        body, resources = html_exporter.from_notebook_node(nb)
        with open(html_path, "w", encoding="utf-8") as f:
            f.write(body)
        print(f"Wrote {html_path}")
    except Exception as e:
        print("Conversion failed:", e)

# Create README.md tailored for Example3
readme = f"""# Example3 — Notebook and Outputs

This folder contains the Example3 Jupyter notebook and exported outputs.

Contents
- {nb_path} — source notebook
- {html_path} — exported HTML version of the notebook (created by this script)
- {zip_path} — archive of common outputs (created if files exist)
- students.csv, students.xlsx, students.pkl — example data files (if present)
- scores_chart.png, plotly_fig.html — example figures (if present)

How to view
- Open {html_path} in a web browser (double-click or use File → Open File).
- In the notebook you can show a clickable link:
  from IPython.display import FileLink, display
  display(FileLink("{html_path}"))

How to re-create the HTML (two options)

1) Using the nbconvert CLI (if installed)
# Run in a notebook cell:
import sys
!{sys.executable} -m pip install --upgrade nbconvert
!{sys.executable} -m jupyter nbconvert --to html {nb_path}

2) Using Python (no external CLI required) — this README's script already does that.

Notes
- nbconvert may warn about missing alt text for images if image markdown is like `![](image.png)`. Add alt text for accessibility: `![description](image.png)`.
"""

with open("README.md", "w", encoding="utf-8") as f:
    f.write(readme)
print("Wrote README.md")

# Create zip of files that exist
with zipfile.ZipFile(zip_path, "w") as z:
    for fn in files_to_include:
        if os.path.exists(fn):
            z.write(fn)
print("Wrote", zip_path, "contains:", zipfile.ZipFile(zip_path).namelist())

# Display links
display(FileLink("README.md"))
if os.path.exists(html_path):
    display(FileLink(html_path))
display(FileLink(zip_path))
```
